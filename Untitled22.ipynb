```python
import pandas as pd
```

```python
df=pd.read_csv('/content/recommend_course1.csv')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

# Load your file (assuming it's already loaded as df)
# df = pd.read_csv('recommend_course.csv')

# --- Step 1: Convert Text to Numbers (Vectorization) ---
# We use TF-IDF to convert the 'name' column into a matrix of numbers.
# stop_words='english' removes common words like "the", "a", "in" which are useless for similarity.
tfidf = TfidfVectorizer(stop_words='english')

# Fill NaN values just in case (though your data looked clean)
df['name'] = df['name'].fillna('')

# Construct the TF-IDF Matrix
tfidf_matrix = tfidf.fit_transform(df['name'])

# --- Step 2: Calculate Similarity ---
# linear_kernel calculates the Cosine Similarity between every course pair.
# It's much faster than cosine_similarity() for this kind of data.
cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)

# --- Step 3: Create a Lookup Mechanism ---
# This ensures if a name appears twice, we only keep the first one
indices = pd.Series(df.index, index=df['name'])
indices = indices[~indices.index.duplicated(keep='first')]

# --- Step 4: Define the Recommendation Function ---
def get_recommendations(title, cosine_sim=cosine_sim):
    if title not in indices:
        return "Course not found in database."

    # Get the index (guaranteed to be a single value now)
    idx = indices[title]

    # If by any chance idx is still a list/array, take the first element
    if hasattr(idx, '__iter__'):
        idx = idx[0]

    # This line will now work without the ValueError
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Get top 5
    sim_scores = sim_scores[1:6]
    course_indices = [i[0] for i in sim_scores]

    return df.iloc[course_indices][['name', 'avg_rating', 'institution']]
```

```python
results = get_recommendations('Machine Learning')
print(results)
```

```python
import gradio as gr

def get_recommendations_html(course_name):
    # 1. Get the recommendations using your existing logic
    if course_name not in indices:
        return "<p style='color:red;'>Course not found. Please select from the dropdown.</p>"

    idx = indices[course_name]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:6] # Top 5

    course_indices = [i[0] for i in sim_scores]
    results = df.iloc[course_indices]

    # 2. Build the HTML Grid
    html_output = "<div style='display: grid; grid-template-columns: repeat(auto-fit, minmax(200px, 1fr)); gap: 20px; padding: 10px;'>"

    for _, row in results.iterrows():
        # HTML for a single Flash Card
        card_html = f"""
        <a href="{row['course_url']}" target="_blank" style="text-decoration: none; color: inherit;">
            <div style="
                border-radius: 15px;
                padding: 20px;
                height: 180px;
                background: white;
                box-shadow: 0 4px 8px rgba(0,0,0,0.1);
                display: flex;
                flex-direction: column;
                justify-content: space-between;
                transition: transform 0.2s;
                border: 1px solid #e0e0e0;
                cursor: pointer;"
                onmouseover="this.style.transform='scale(1.05)'"
                onmouseout="this.style.transform='scale(1)'">

                <div style="font-size: 1.1em; font-weight: bold; line-height: 1.3; color: #2D3E50;">
                    {row['name']}
                </div>

                <div style="display: flex; justify-content: space-between; align-items: center; margin-top: 10px;">
                    <span style="font-size: 0.8em; color: #7f8c8d;">{row['institution']}</span>
                    <span style="background: #f1c40f; padding: 2px 8px; border-radius: 5px; font-size: 0.8em; font-weight: bold;">
                        ★ {row['avg_rating']}
                    </span>
                </div>
            </div>
        </a>
        """
        html_output += card_html

    html_output += "</div>"
    return html_output

# 3. Create the Gradio Interface
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🎓 Smart Course Recommender")
    gr.Markdown("Select a course you liked, and we'll suggest the next 5 best matches.")

    with gr.Row():
        course_input = gr.Dropdown(
            choices=list(df['name'].unique()),
            label="Pick a Course",
            interactive=True
        )

    with gr.Row():
        # This HTML component will act as our "Gallery" of flashcards
        recommendation_output = gr.HTML(label="Recommended Courses")

    # Connect the logic
    course_input.change(fn=get_recommendations_html, inputs=course_input, outputs=recommendation_output)

# 4. Launch
demo.launch(debug=True)
```
