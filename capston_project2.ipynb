```python
def create_url(row):
  a=row['course'].lower().split()
  aa=("-").join(a)
  if row['certificatetype'] in ['Professional Certificate','Specialization']:
    return "https://www.coursera.org/"+('-').join(row['certificatetype'].lower().split())+"/"+aa
  elif row['certificatetype']=='Course':
    return "https://www.coursera.org/"+"learn"+"/"+aa
  else:
    return "https://www.coursera.org/"+"project"+"/"+aa

# print(create_url({'certificatetype':'Professional Certificate', 'course':'google Cyberseceuty'}))
```

```python
df1['course_url']=df1.apply(create_url,axis=1)
```

```python
df1=df1.drop(['skills','certificatetype','duration'],axis=1)
```

```python
def new_rating(a):
  try:
    aa=float(a[:-1].strip())
    if aa>0 and aa<15:
      return 1
    elif aa>=15.0 and aa<35:
      return 2
    elif aa>=35.0 and aa<50.0:
      return 3
    elif aa>=50.0 and aa<100.0:
      return 4
    else:
      return 5
  except:
    return 1


df1['rating_count']=df1['reviewcount'].apply(new_rating)
```

```python
df1['additional_rating']= df1['crediteligibility'].apply(lambda x :0.1 if x==True else 0)
```

```python
df1['final_rating'] = df1['rating']*0.7+df1['rating_count']*0.2+df1['additional_rating']*0.5
```
